# **MULTIVARIATE ANALYSIS**

```python
tips = sns.load_dataset('tips')
```

```python
tips.sample(5)
```

### 1) Scatter Plot (Numerical-Numerical)

```python
plt.figure(figsize=(10,7))
sns.scatterplot(x="total_bill",y="tip",data=tips,hue="sex",style="smoker",size="size")
plt.show()
```

### 2) Bar Plot (Numerical-Categorical)

```python
sns.barplot(x="Pclass",y="Fare",data=data, hue="Sex")
plt.show()
```

### 3) Box Plot (Numerical-Categorical)

```python
sns.boxplot(x="Sex",y="Age",data=data, hue="Survived")
plt.show()
```

### 4) Dist Plot (Numerical-Categorical)

```python
sns.kdeplot(
    data=data[data["Survived"] == 0],
    x="Age",
    label="Not Survived"
)

sns.kdeplot(
    data=data[data["Survived"] == 1],
    x="Age",
    label="Survived"
)
plt.legend()
plt.show()
```

### 5) Heat Plot (Categorical-Categorical)

```python
sns.heatmap(pd.crosstab(data["Pclass"],data["Survived"]))
plt.show()
```

### 6) Pair Plot

```python
data = sns.load_dataset("iris")
```

```python
sns.pairplot(data,hue="species")
plt.show()
```
