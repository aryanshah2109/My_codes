# Scatter Plots

image.png

- Bivariate Analysis
- numerical vs numerical
- Use case - Finding correlation

```python
age = [13,19,23,20,21]
watchtime = [5,9,6,3,8]

plt.scatter(age, watchtime)
plt.title("Age vs TV Watchtime")
plt.xlabel("Age")
plt.ylabel("Watchtime")
plt.show()
```

```python
x = np.linspace(-10,10,50)
y = 10 * x + 3 + np.random.randint(0,600,50)

plt.scatter(x,y)
plt.show()
```

```python
# from dataset
score = pd.read_csv("batter.csv")
score = score.sample(50)
score.sample(5)
```

```python
# Avg vs Strike rate of random 50 batsmen
# markersize = s in scatter plot
plt.scatter(score.avg, score.strike_rate, color="red", marker="+", s=200)
plt.title("Average vs Strike rate")
plt.xlabel("Average")
plt.ylabel("Strike rate")
plt.show()
```

```python
# We can set 's' attribute to a particular column in dataset
# That will lead to different sized markers based on the value of the chosen column

tips = sns.load_dataset("tips")
print(tips.sample(5))

print("\n\n")

plt.scatter(tips["total_bill"], tips["tip"], s=tips["size"]*15)
plt.title("Bill VS Tips")
plt.xlabel("Bill")
plt.ylabel("Tips")
plt.show()
```

```python
# scatter plot with plt.plot()
# cannot manipulate size and color like in plt.scatter()
# faster than plt.scatter()

plt.plot(tips["total_bill"], tips["tip"], "o")
plt.title("Bill VS Tips using plot function")
plt.xlabel("Bill")
plt.ylabel("Tips")
plt.show()
```

```python
# simple bar chart
children = [10,20,40,10,30]
colors = ['red','blue','green','yellow','pink']

plt.bar(colors,children,color='orange')
```

```python
# horizontal bar chart using barh

children = [10,20,40,10,30]
colors = ['red','blue','green','yellow','pink']

plt.barh(colors,children,color='orange')
```

```python
df = pd.read_csv('/content/batsman_season_record.csv')
df
```

```python
# multi column bar chart
# First make x label numerical to perform shifting of bars by adding/subtracting
plt.bar(np.arange(df.shape[0]) - 0.2,df['2015'],width=0.2,color='yellow')
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.2,color='red')
plt.bar(np.arange(df.shape[0]) + 0.2,df['2017'],width=0.2,color='blue')

# Then convert current numerical x labels to actual categorical x labels
plt.xticks(np.arange(df.shape[0]), df['batsman'])

plt.show()
```

```python
# if your category names are too large for x axis, rotate them
children = [10,20,40,10,30]
colors = ['red red red red red red','blue blue blue blue','green green green green green','yellow yellow yellow yellow ','pink pinkpinkpink']

plt.title("Without rotation")
plt.bar(colors, children)
plt.show()

plt.title("With rotation")
plt.bar(colors, children)
plt.xticks(rotation="vertical")
plt.show()
```

```python
# stacked bar charts
plt.bar(df["batsman"], df["2015"],label="2015")
plt.bar(df["batsman"], df["2016"], bottom=df["2015"],label="2016")
plt.bar(df["batsman"], df["2017"], bottom=df["2016"]+df["2015"],label="2017")
plt.legend()
```

### Colored Scatterplots

```python
iris = pd.read_csv("iris.csv")
iris.head()
```

```python
plt.scatter(iris["SepalLengthCm"],iris["PetalLengthCm"])
plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")
plt.show()
```

```python
# To multi color, convert species(categorical) to numerical
from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()
iris["Species"] = encoder.fit_transform(iris["Species"])

# alpha -> opacity, c -> different colors, cmap -> color of colors, colorbar->vertical bar with intensity values
plt.scatter(iris["SepalLengthCm"],iris["PetalLengthCm"], c=iris["Species"], cmap="jet", alpha=0.9)
plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")
plt.colorbar()
plt.show()
```

### Plot size

```python
plt.figure(figsize=(10,5))
plt.scatter(iris["SepalLengthCm"],iris["PetalLengthCm"], c=iris["Species"], cmap="jet", alpha=0.9)
plt.show()
```

### Subplots

```python
# Alernative way to plot graphs

fig, ax = plt.subplots()

ax.scatter(batters['avg'],batters['strike_rate'])
ax.set_title("Avg VS SR")
ax.set_xlabel("Avg")
ax.set_ylabel("SR")

fig.show()
```

```python
# 2 graphs : (avg vs runs) and (avg vs strike_rate)

fig, ax = plt.subplots(nrows=2,ncols=1, figsize=(10,12), sharex=True)

ax[0].scatter(batters["avg"],batters["runs"], color="orange")
ax[1].scatter(batters["avg"],batters["strike_rate"], color="green")

ax[0].set_title("Avg VS Runs")
ax[0].set_xlabel("Avg")

ax[1].set_title("Avg VS SR")
ax[1].set_xlabel("Avg")
ax[1].set_ylabel("SR")

fig.show()
```

```python
# 4 graphs : (avg vs runs), (avg vs strike_rate), histogram(avg), histogram(runs)

fig, ax = plt.subplots(nrows=2,ncols=2, figsize=(10,12))

ax[0,0].scatter(batters["avg"],batters["runs"], color="orange")
ax[0,1].scatter(batters["avg"],batters["strike_rate"], color="green")
ax[1,0].hist(batters["avg"],color="brown")
ax[1,1].hist(batters["runs"],color="red")


fig.show()
```

```python
# Alternate way
fig = plt.figure()

ax1 = fig.add_subplot(2,2,1)
ax1.scatter(batters["avg"],batters["runs"], color="orange")

ax2 = fig.add_subplot(2,2,2)
ax2.scatter(batters["avg"],batters["strike_rate"], color="green")

ax3 = fig.add_subplot(2,2,3)
ax3.hist(batters["avg"],color="brown")

ax4 = fig.add_subplot(2,2,4)
ax4.hist(batters["runs"],color="red")

fig.show()
```

### 3D Surface Plots

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
```

```python
xx, yy = np.meshgrid(x,y)
```

```python
z = xx**2 + yy**2
```

```python
fig = plt.figure(figsize=(20,8))

ax = plt.subplot(projection="3d")

ax.plot_surface(xx,yy,z,cmap="Blues")

fig.show()
```
