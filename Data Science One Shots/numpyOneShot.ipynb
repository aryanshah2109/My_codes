## What is NumPy?

NumPy is the core library for numerical computing in Python. It provides the `ndarray` object and a large collection of functions for fast computation on arrays.

```python
import numpy as np
import matplotlib.pyplot as plt

np.set_printoptions(suppress=True)
```

## Creating NumPy Arrays

This section covers `np.array`, `np.arange`, `np.ones`, `np.zeros`, `np.random.random`, `np.linspace`, and `np.identity`.


```python
# 1D, 2D, and 3D arrays
a = np.array([1, 2, 3])
b = np.array([[1, 2, 3], [4, 5, 6]])
c = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])

print(a)
print(b)
print(c)

# Different dtypes
print(np.array([1, 2, 3], dtype=float))
print(np.array([1, 2, 3], dtype=bool))

# Initializing with particular values
print(np.arange(1, 11, 2))
print(np.arange(16).reshape(2, 2, 2, 2))
print(np.ones((3, 4)))
print(np.zeros((3, 4)))
print(np.random.random((3, 4)))
print(np.linspace(-10, 10, 10, dtype=int))
print(np.identity(3))
```

## Array Attributes

Inspect dimensions, shape, size, item size, and datatype.


```python
a1 = np.arange(10, dtype=np.int32)
a2 = np.arange(12, dtype=float).reshape(3, 4)
a3 = np.arange(8).reshape(2, 2, 2)

print(a1)
print(a2)
print(a3)

print(a1.ndim)
print(a2.ndim)
print(a3.ndim)
print(a3.shape)
print(a2.size)
print(a3.itemsize)
print(a1.dtype)
print(a2.dtype)
print(a3.dtype)
```

## Changing Datatype

Use `astype` to convert arrays to a new datatype.


```python
print(a3.astype(np.int32))
```

## Array Operations

NumPy supports elementwise arithmetic, comparison, and vectorized operations.


```python
a1 = np.arange(12).reshape(3, 4)
a2 = np.arange(12, 24).reshape(3, 4)

print(a2)
print(a1 ** 2)
print(a2 > 15)
print(a2 == 15)
print(a1 ** a2)
```

## Array Functions

This section shows common reductions, trigonometric functions, matrix multiplication, and exponential/rounding helpers.


```python
a1 = np.random.random((3, 3))
a1 = np.round(a1 * 100)
print(a1)

print(np.max(a1))
print(np.min(a1))
print(np.sum(a1))
print(np.prod(a1))
print(np.prod(a1, axis=0))
print(np.mean(a1))
print(np.median(a1))
print(np.std(a1))
print(np.var(a1))
print(np.sin(a1))

a2 = np.arange(12).reshape(3, 4)
a3 = np.arange(12, 24).reshape(4, 3)
print(np.dot(a2, a3))
print(np.exp(a1))
print(np.ceil(np.random.random((2, 3)) * 100))
```

## Indexing and Slicing

Examples for 1D, 2D, and 3D indexing and slicing.


```python
print("\nIndexing\n")

a1 = np.arange(10)
a2 = np.arange(12).reshape(3, 4)
a3 = np.arange(8).reshape(2, 2, 2)

print(a3)
print(a1)
print(a2)
print(a1[3])
print(a2[1, 0])
print(a3[1, 0, 1])
print(a3[1, 1, 0])
print(a1[2:5:2])
print(a2[0:2, 1::2])
print(a2[::2, 1::2])
print(a2[1, ::3])
print(a2[0, :])
print(a2[:, 2])
print(a2[1:, 1:3])

a3 = np.arange(27).reshape(3, 3, 3)
print(a3)
print(a3[::2, 0, ::2])
print(a3[2, 1:, 1:])
print(a3[0, 1, :])
```

## Iterating

Iterate through arrays directly and with `np.nditer`.


```python
print("\n\nLooping")
print(a1)
for i in a1:
    print(i)

print(a2)
for i in a2:
    print(i)

print(a3)
for i in a3:
    print(i)

for i in np.nditer(a3):
    print(i)
```

## Transpose and Ravel

`transpose`/`.T` swap axes, and `ravel` flattens the array.


```python
print("\n\nTranspose")
print(np.transpose(a2))
print(a2.T)
print(a3.ravel())
```

## Stacking

Horizontal and vertical stacking with `hstack` and `vstack`.


```python
a4 = np.arange(12).reshape(3, 4)
a5 = np.arange(12, 24).reshape(3, 4)

print(a5)
print(np.hstack((a4, a5)))
print(np.vstack((a4, a5)))
```

## Splitting

Split arrays horizontally and vertically.


```python
print(a4)
print(np.hsplit(a4, 2))

print(a5)
print(np.vsplit(a5, 3))
```

## Advanced NumPy

Fancy indexing, boolean indexing, broadcasting, missing values, plotting, and utility helpers.


```python
print("\n\nArray: ")
arr = np.random.randint(1, 100, 35).reshape(7, 5)
print(arr)

print("\n\n Fancy Indexing")
print(arr[[3, 4, 5]])
print("\n")
print(arr[[1, 5, 2]])

print("\n\n Boolean Indexing")
arr = np.random.randint(1, 100, 25).reshape(5, 5)
print(arr)
print("\n")
print(arr[arr > 70])
print("\n")
print(arr[(arr % 2 == 0) & (arr > 50)])

print("\nBroadcasting::")
arr1 = np.array([[1, 2, 3], [4, 5, 6]])
arr2 = np.array([1, 2, 3])
print(arr1 + arr2)

arr1 = np.arange(12).reshape(3, 4)
arr2 = np.arange(4)
print(arr1)
print(arr2)
print(arr1 + arr2)

arr = np.array([1, 2, 3, 4, np.nan, 5, 6])
print(arr)
print(arr[~np.isnan(arr)])
```

## Plotting Mathematical Functions

A few quick graphs using Matplotlib and NumPy.


```python
fig, axes = plt.subplots(2, 3, figsize=(12, 7))
axes = axes.ravel()

x = np.linspace(-10, 10, 100)
plots = [
    (x, x, "y = x"),
    (x, x ** 2, "y = x^2"),
    (x, np.sin(x), "y = sin(x)"),
    (x, np.exp(x), "y = exp(x)"),
    (x, x ** 3, "y = x^3"),
]

for ax, (xv, yv, title) in zip(axes, plots):
    ax.plot(xv, yv)
    ax.set_title(title)
    ax.grid(True, alpha=0.3)

axes[-1].axis("off")
plt.tight_layout()
plt.show()
```

## Extra Functions

Sorting, appending, concatenation, unique values, dimension expansion, `where`, and `argmax`.


```python
print("\n\n Sorting arrays")
arr = np.random.randint(1, 50, 15)
print("\n")
print(arr)
print("\n")
print(np.sort(arr))

arr = np.random.randint(1, 50, 15).reshape(3, 5)
print("\n")
print(arr)
print("\n")
print(np.sort(arr))
print("\n")
print(np.sort(arr, axis=0))

print("\n\n Appending at end in arrays")
arr = np.random.randint(1, 50, 15)
print("\n")
print(arr)
print("\n")
print(np.append(arr, 250))

arr = np.random.randint(1, 50, 15).reshape(3, 5)
print("\n")
print(arr)
print("\n")
print(np.append(arr, np.ones((arr.shape[0], 1)), axis=1))

print("\n\nConcatenate arrays:")
arr = np.random.randint(1, 50, 15).reshape(3, 5)
print(arr)
print("\n")
brr = np.random.randint(1, 50, 15).reshape(3, 5)
print(brr)
print("\n")
crr = np.concatenate((arr, brr), axis=1)
print(crr)
print("\n")
crr = np.concatenate((arr, brr), axis=0)
print(crr)
print("\n")

print("\n\nUnique elements:")
arr = np.array([1, 1, 3, 3, 4, 5, 6, 3, 4, 5, 2, 5, 6, 3, 4])
print("\n")
print(arr)
print("\n")
print(np.unique(arr))

print("\n\nExpand dims:")
arr = np.random.randint(1, 100, 15)
print("\n")
print(arr)
print("\n")
print(np.expand_dims(arr, axis=0))
print("\n")
print(np.expand_dims(arr, axis=1))

print("\n\nWhere function:")
arr = np.random.randint(1, 100, 15)
print("\n")
print(arr)
print("\n")
print(np.where(arr > 50))
print("\n")
print(np.where(arr < 30, 0, arr))

print("\n\nArgmax function:")
print(arr)
print(np.argmax(arr))
print("\nSimilarly, np.argmin(arr) returns the index of the minimum element.")
```
