```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# **Dates:**

```python
data = pd.read_csv("orders.csv")
data.sample(5)
```

```python
data.info()
```

```python
data["date"] = pd.to_datetime(data["date"])
```

```python
data.info()
```

```python
data.sample(5)
```

```python
# Extract year, month and day
data["date_year"] = data["date"].dt.year
data["date_month"] = data["date"].dt.month
data["date_day"] = data["date"].dt.day

data.sample(5)
```

```python
# Extract month name
data["date_month_name"] = data["date"].dt.month_name()

data.sample(5)
```

```python
# Extract time elapsed between todays date and data's dates

import datetime

dt_current = datetime.datetime.today()

dt_current
```

```python
dt_current - data["date"]
```

```python
(dt_current - data["date"]).dt.days
```
