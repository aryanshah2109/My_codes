```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
```

```python
data = pd.read_csv("/content/data/creditcard.csv")
data = data.sample(5000)
```

```python
X_train, X_test, y_train, y_test = train_test_split(data.drop("Class", axis=1), data["Class"], test_size=0.2, random_state=32)
```

```python
model = RandomForestClassifier()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
```

```python
print(accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

# **TECHNIQUES**

1) Undersampling

```python
from imblearn.under_sampling import RandomUnderSampler
```

```python
data = pd.read_csv("/content/data/creditcard.csv")
data.shape
```

```python
data["Class"].value_counts()
```

```python
X = data.drop("Class", axis=1)
y = data["Class"]
```

```python
X_sampled, y_sampled = RandomUnderSampler(random_state=32).fit_resample(X, y)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_sampled, y_sampled, test_size=0.2, random_state=32)
```

```python
model = LogisticRegression()
model.fit(X_sampled, y_sampled)
y_pred = model.predict(X_test)
```

```python
print(accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

```python
X_sampled.shape
```

```python
y_sampled.value_counts()
```

4) Class weights

```python
data = pd.read_csv("/content/data/creditcard.csv")
data.shape
```

```python
data["Class"].value_counts()
```

```python
X = data.drop("Class", axis=1)
y = data["Class"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=32)
```

```python
model = LogisticRegression(class_weight={0:1, 1:5}, solver='liblinear')
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
```

```python
print(accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
```
