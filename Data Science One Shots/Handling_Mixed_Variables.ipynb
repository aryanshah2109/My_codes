```python
# Splitting number into number_numerical and number_categorical

data["number_numerical"] = pd.to_numeric(data["number"], errors="coerce", downcast="integer")

data["number_categorical"] = np.where(data["number_numerical"].isna(), data["number"], np.nan)
```
