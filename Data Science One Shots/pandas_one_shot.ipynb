### Pandas Series

A Pandas Series is like a column in a table. It is a 1-D array holding data of any type.

```python
import numpy as np
import pandas as pd
```

### 1) Creating Series

```python
# Series from list of integers
series1 = pd.Series([1,2,3,4,5,6])
print(series1)
```

```python
# Series from list of strings
series2 = pd.Series(["Apples", "Oranges", "Kiwi", "Bananas"])
print(series2)
```

```python
# Custom indices and names
marks = [10,20,30,40,50]
students = ["A", "B", "C", "D", "E"]
series3 = pd.Series(marks, index=students, name="Student marks")
print(series3)
```

```python
# Series from dictionary
marks = {
    "A": 10,
    "B":20,
    "C":30,
    "D":40,
    "E":50,
}
series = pd.Series(marks)
print(series)
```

### 2) Series Attributes

```python
marks = {
    "A": 10,
    "B":20,
    "C":30,
    "D":40,
    "E":50,
}
series = pd.Series(marks, name="Marks of Students")
print(series)

series1 = pd.Series([1,2,3,4,5,6])
series2 = pd.Series([1,2,2,4,2,6])
print("\n")
print(series1)
print("\n")
print(series2)
```

### 3) Series Methods

```python
series = pd.Series(np.random.randint(1,100,30))
print(series)
```

```python
# head() and tail() : head(n) -> top n rows, tail(n) -> bottom n rows
# if n is not passed, by default, 5 rows are returned
print(series.head())
print("\n")
print(series.head(10))
print("\n")
print(series.tail())
print("\n")
print(series.tail(10))
print("\n")
```

```python
# sample(n) -> return random n rows
# if n is not passed, by default, 1 row are returned
print(series.sample())
print("\n")
print(series.sample(10))
print("\n")
```

```python
# value_counts() -> returns number of occurances of each value
print(series.value_counts())
```

```python
# sort_values() -> sorts series

# Ascending order
print(series.sort_values())
print("\n\n")
# Descending order
print(series.sort_values(ascending=False))
```

```python
# inplace parameter : Modifies original series
series.sort_values(inplace=True)
print(series)
```

```python
# sort_index() : sorts indices
# parameters: ascending = True/False , inplace = True/False
series.sort_index(inplace=True)
print(series)
```

### 4) Series Math Methods

```python
series = pd.Series(np.random.randint(1,100,10))
print(series)
```

```python
# count : returns number of elements in series
# similar to size but in size, it includes missing values
# but count doesnt

series1 = pd.Series([1,2,4,3,np.nan,8,9,11])

print(series1.count())
print(series1.size)
```

### 5) Series Indexing and Slicing

```python
series1 = pd.Series(np.random.randint(1,100,20))
print(series1)

print("\n\n")

series2 = pd.Series(["Apple","Banana","Grape","Plum","Orange","Kiwi","Guava","Watermelon"])
print(series2)
```

```python
# Integer indexing
print(series1[0])
print(series2[5])
```

```python
# Slicing
print(series1[0:5])
print("\n")

print(series1[-5:])
print("\n")

print(series1[2::2])
print("\n")

print(series2[0:5])
print("\n")

print(series2[-5:])
print("\n")
```

```python
# Fancy Indexing
print(series1[[1,4,6,9,12]])
print("\n")
print(series2[[1,4,6]])
```

```python
# Boolean Indexing
print(series1[series1>50])
```

### 6) Excess Functions Series

```python
series = pd.Series([1,1,0,0,1,1,0,1])
series
```

```python
# astype: changes datatype of series
series = series.astype(bool)
series
```

```python
# drop_duplicates: removes duplicates from series
series = pd.Series([1,2,3,1,1,3,4,1,3,5,6,1,4,5,9])
series.drop_duplicates()
```

```python
# duplicated: creates a series of bools
# Can be used to find how many duplicates in series
series = pd.Series([1,2,3,1,1,3,4,1,3,5,6,1,4,5,9])
series.duplicated().sum()
```

```python
# isnull : creates a series of bools
# Can be used to find how many nulls in series
series = pd.Series([1,4,np.nan,5,np.nan,1,9,3,np.nan])
series.isna().sum()
```

```python
# dropna : deletes all missing values
print(series)
print("\n")
temp = series.dropna()
print(temp)
```

```python
# fillna : fills all missing values with given values
print(series)
print("\n")
temp = series.fillna("Missing")
print(temp)
```

```python
# apply : allows using functions on each row of series
def add(s):
  return s+10

series = pd.Series([1,2,3,4,5,6,7])
print(series.apply(add))
```

### 4) Accessing rows from DF

```python
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eve", "Frank"],
    "Age": [24, 27, 22, 25, 26, 28],
    "City": ["Delhi", "Mumbai", "Bangalore", "Chennai", "Pune", "Hyderabad"],
    "Salary": [50000, 60000, 55000, 58000, 62000, 65000],
    "Experience": [2, 4, 1, 3, 5, 6]
}

index = [
    "id_101",
    "id_102",
    "id_103",
    "id_104",
    "id_105",
    "id_106"
]

df = pd.DataFrame(data, index=index)
print(df)
```

```python
# iloc: Searched used index positions
print(df.iloc[0])   # Single row
print("\n")
print(df.iloc[0:5])  # Multiple rows
print("\n")
print(df.iloc[[0,4,1,3]])  # Fancy indexing
```

```python
# loc : Fetches rows according to index name
print(df.loc["id_101"])   # Single row
print("\n")
print(df.loc["id_101":"id_104"])  # Multiple rows
print("\n")
print(df.loc[["id_101","id_104","id_103"]])  # Fancy indexing
```

```python
# create a function that will take names of 2 ipl teams and return how many matches each won against each other

def team_win_filter(df, team1, team2):
  both_teams = df[
      (
          (df["Team1"] == team1) & (df["Team2"] == team2)
      )
            |
      (
          (df["Team1"] == team2) & (df["Team2"] == team1)
      )
    ]

  return both_teams["WinningTeam"].value_counts()

team1 = "Royal Challengers Bangalore"
team2 = "Chennai Super Kings"

df = team_win_filter(ipl, team1, team2)
df
```
